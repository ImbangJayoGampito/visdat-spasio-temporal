```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
def extract_dataframe(filename, filetype = 'csv'):
    if filetype == 'csv':
        df = pd.read_csv(filename)
    elif filetype == 'xlsx':
        df = pd.read_excel(filename)
    return df
```

```python
df1 = extract_dataframe('Persentase Penduduk Miskin Menurut Kabupaten_Kota di Sumatera Barat, 2006-2025.xlsx', 'xlsx')
# Change row 1 to be int else make it as nan

df1
```

```python
df2 = extract_dataframe('Data_Pendapatan_Rumah_Tangga_Menurut_Kabupaten_Kota.csv', 'csv')
df2
```

```python
def process_dataframe(df, year_start=2015, year_end=2025, row_start=2, row_end=2, regency_name = 'Kabupaten/Kota', jenis_data = ['Umur Harapan Hidup (Tahun)', 'Harapan Lama Sekolah (Tahun)','Rata-rata Lama Sekolah (Tahun)','Pengeluaran (Ribu rupiah/orang/tahun)', 'IPM']):
    row_range = (row_start, row_end)
    year_range = (year_start, year_end)
    df_processed = df.copy()
    year_idx = 1
    current_data_index = None
    for column in df.columns:
        year_val = df[column].iloc[year_idx]
        print(year_val, current_data_index, year_start, year_start == year_val and current_data_index is None)
        if pd.isna(year_val) or year_val is None:
            continue
        year_val = int(year_val)
        if year_val == year_start and current_data_index is None:
            print("meowmeow")
            current_data_index = 0
        elif year_val == year_start and current_data_index is not None:
            current_data_index += 1
        
        if current_data_index is not None and current_data_index < len(jenis_data):
            renamed_column_name = f"{jenis_data[current_data_index]}_{int(year_val)}"
            print(f"Renaming {column} to {renamed_column_name}")
            df_processed.rename(columns={column: renamed_column_name}, inplace=True)

    # Remove first two and last two rows which are not needed
    if row_end > 0:
        df_processed = df_processed.iloc[row_start:-row_end].copy()
    else:
        df_processed = df_processed.iloc[row_start:].copy()
        
    # Remove  Kab. initial
    

 
    

    return df_processed
```

```python
df_proc2 = process_dataframe(df1, year_start=2006, year_end=2023, row_start=3, row_end=0, regency_name='Wilayah', jenis_data=['Penduduk_Miskin_(%)'])
df_proc1 = process_dataframe(df2, year_start=2000, year_end=2025, row_start=2, row_end=3, regency_name='Kabupaten/Kota', jenis_data=['Rumah_Tangga_(Ruta)', 'Anggota_Rumah_Tangga_(ART)'])
df_proc1
```

```python
df_proc2['Wilayah'] = df_proc2['Wilayah'].str.replace('Kab.', '', regex=False)
df_proc1['Kabupaten/Kota'] = df_proc1['Kabupaten/Kota'].str.replace('Kab.', '', regex=False)
mask = df_proc1['Kabupaten/Kota'] != 'Kota Solok'
df_proc1.loc[mask, 'Kabupaten/Kota'] = df_proc1.loc[mask, 'Kabupaten/Kota'].str.replace('Kota', '', regex=False)
df_proc1['Kabupaten/Kota'] = df_proc1['Kabupaten/Kota'].replace('Lima Puluh', 'Lima Puluh Kota')
df_proc2['Wilayah'] = df_proc2['Wilayah'].str.strip()
df_proc1['Kabupaten/Kota'] = df_proc1['Kabupaten/Kota'].str.strip()
```

```python
# Remove first row from df_proc2 which is the header row
df_proc2_first_gone = df_proc2.iloc[1:].reset_index(drop=True)
# Merge the two processed dataframes on the regency name column
merged_df = pd.merge(
    df_proc1, 
    df_proc2_first_gone, 
    left_on='Kabupaten/Kota', 
    right_on='Wilayah', 
    how='left'
)
# Get unique names from both datasets
names_proc1 = df_proc1['Kabupaten/Kota'].unique().tolist()
names_proc2 = df_proc2_first_gone['Wilayah'].unique().tolist()

print("Names in dataset1 (df_proc1):")
print(names_proc1)
print("\nNames in dataset2 (df_proc2_first_gone):")
print(names_proc2)


matches = set(names_proc1) & set(names_proc2)
print(f"\nExact matches: {matches}")
print(f"Number of exact matches: {len(matches)}")

# Merge the two processed dataframes on the regency name column
merged_df = merged_df.drop('Wilayah', axis=1)
print(f"Number of rows after merge: {len(merged_df)}")
print(f"Columns: {merged_df.columns.tolist()}")
merged_df
```

```python
# Initialize dictionary with categories as keys

final_dict = {
    'Kabupaten/Kota': [],
    'Tahun': [],
    'Umur Harapan Hidup (Tahun)': [],
    'Harapan Lama Sekolah (Tahun)': [],
    'Rata-rata Lama Sekolah (Tahun)': [],
    'Pengeluaran (Ribu rupiah/orang/tahun)': [],
    'IPM': []
}
for idx, row in merged_df.iterrows():
    regency = row['Kabupaten/Kota']
    year_range = range(2015, 2026) 
    
    # Get columns for each category (sorted by year)
    uhh_cols = sorted([col for col in merged_df.columns if col.startswith('Umur Harapan Hidup (Tahun)_')])
    hls_cols = sorted([col for col in merged_df.columns if col.startswith('Harapan Lama Sekolah (Tahun)_')])
    rls_cols = sorted([col for col in merged_df.columns if col.startswith('Rata-rata Lama Sekolah (Tahun)_')])
    pengeluaran_cols = sorted([col for col in merged_df.columns if col.startswith('Pengeluaran (Ribu rupiah/orang/tahun)_')])
    ipm_cols = sorted([col for col in merged_df.columns if col.startswith('IPM_')])
    
    # Get values
    uhh_values = row[uhh_cols].values
    hls_values = row[hls_cols].values
    rls_values = row[rls_cols].values
    pengeluaran_values = row[pengeluaran_cols].values
    ipm_values = row[ipm_cols].values
    
    # Loop through each year
    for i, year in enumerate(year_range):
        final_dict['Kabupaten/Kota'].append(regency)
        final_dict['Tahun'].append(year)
        final_dict['Umur Harapan Hidup (Tahun)'].append(uhh_values[i] if i < len(uhh_values) else None)
        final_dict['Harapan Lama Sekolah (Tahun)'].append(hls_values[i] if i < len(hls_values) else None)
        final_dict['Rata-rata Lama Sekolah (Tahun)'].append(rls_values[i] if i < len(rls_values) else None)
        final_dict['Pengeluaran (Ribu rupiah/orang/tahun)'].append(pengeluaran_values[i] if i < len(pengeluaran_values) else None)
        final_dict['IPM'].append(ipm_values[i] if i < len(ipm_values) else None)


final_df = pd.DataFrame(final_dict)

print(f"Shape: {final_df.shape}")
print(f"Expected: ~20 regencies × 11 years = 220 rows")
print(f"Actual: {len(final_df)} rows")
print("\nFirst 10 rows:")
print(final_df.head(10))

```
